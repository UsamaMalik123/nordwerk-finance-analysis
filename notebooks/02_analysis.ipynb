```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.float_format', '{:,.0f}'.format)
```

```python
gl=pd.read_csv("gl_clean.csv", parse_dates=['posting_date'])
```

```python
coa = pd.read_csv("coa_clean.csv")
```

```python
gl.info()
```

```python
df=gl.merge(coa,on='account_no',how='left')
```

```python
df.head()
```

```python
df['account_group'].isna().sum()
```

```python
df['account_group']=df['account_group'].fillna('Unmapped')
```

```python
df[df['account_group'] == "Unmapped"]
```

```python
df.info()
```

```python
df['year']=df['posting_date'].dt.year
```

```python
df.head(3)
```

```python
df['Month']=df['posting_date'].dt.month
```

```python
df.head(3)
```

```python
df[['posting_date', 'year', 'Month']].head()
```

```python
pnl = df.pivot_table(index='account_group', columns='year', values='amount', aggfunc='sum')
pnl
```

```python
revenue = pnl.loc['Revenue']
```

```python
revenue
```

```python
###cogs is cost of goods
gross_profit = revenue + pnl.loc['COGS']
```

```python
gross_profit
```

```python
##earning before interest and taxes
ebit = gross_profit + pnl.loc['Personnel'] + pnl.loc['OPEX']
ebit
```

```python
gross_margin = gross_profit / revenue * 100
ebit_margin = ebit / revenue * 100
```

```python
summary = pd.DataFrame({'Revenue': revenue, 'Gross Profit': gross_profit, 'Gross Margin %': gross_margin, 'EBIT': ebit, 'EBIT Margin %': ebit_margin})
summary
```

```python
# ### Finding 1: Profit halved despite higher sales
# - Revenue grew 4% (€10.9M → €11.4M)
# - But EBIT fell 52% (€1.30M → €0.62M); the EBIT margin dropped from 12% to 5%
# - Gross margin fell from 57% to 54%, so material costs (COGS) grew faster than sales (+13%)
# - OPEX also grew 14%
# - Next question: which costs exactly? → Budget vs Actual analysis
    
```

```python
summary.to_csv("pnl_summary.csv")
```
